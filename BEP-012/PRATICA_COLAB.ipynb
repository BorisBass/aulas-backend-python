## 1. Configuração Inicial

Primeiro, vamos importar o módulo sqlite3 e criar um banco de dados em memória.


```python
import sqlite3

# Criar conexão com banco em memória
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()
print("✅ Conexão com SQLite estabelecida!")
```

## 2. DDL - Criar Tabelas (CREATE)

Vamos criar as tabelas para nossa prática.


```python
# Criar tabela de clientes
cursor.execute('''
CREATE TABLE clientes (
    id INTEGER PRIMARY KEY,
    nome TEXT NOT NULL,
    email TEXT,
    telefone TEXT,
    cidade TEXT,
    data_cadastro DATE
)
''')

# Criar tabela de produtos
cursor.execute('''
CREATE TABLE produtos (
    id INTEGER PRIMARY KEY,
    nome TEXT NOT NULL,
    preco REAL,
    estoque INTEGER
)
''')

print("✅ Tabelas criadas com sucesso!")
```

## 3. DML - Inserir Dados (INSERT)

Agora vamos inserir dados de exemplo nas tabelas.


```python
# Inserir clientes
clientes = [
    ('João Silva', 'joao@email.com', '(71) 99999-1111', 'Salvador', '2024-01-15'),
    ('Maria Santos', 'maria@email.com', '(71) 99999-2222', 'Feira de Santana', '2024-01-20'),
    ('Pedro Costa', 'pedro@email.com', '(71) 99999-3333', 'Salvador', '2024-01-25'),
    ('Ana Lima', 'ana@email.com', '(71) 99999-4444', 'Vitória da Conquista', '2024-02-01'),
    ('Carlos Oliveira', 'carlos@email.com', '(71) 99999-5555', 'Camaçari', '2024-02-05')
]

cursor.executemany('''
INSERT INTO clientes (nome, email, telefone, cidade, data_cadastro)
VALUES (?, ?, ?, ?, ?)
''', clientes)

print(f"✅ {len(clientes)} clientes inseridos!")
```

```python
# Inserir produtos
produtos = [
    ('Notebook Dell', 2500.00, 15),
    ('Mouse Logitech', 89.90, 50),
    ('Teclado Mecânico', 199.90, 25),
    ('Monitor 24"', 800.00, 10),
    ('Webcam HD', 150.00, 30)
]

cursor.executemany('''
INSERT INTO produtos (nome, preco, estoque)
VALUES (?, ?, ?)
''', produtos)

print(f"✅ {len(produtos)} produtos inseridos!")
```

## 4. DML - Consultar Dados (SELECT)

Agora vamos consultar os dados inseridos.


```python
# SELECT * - Todos os clientes
cursor.execute("SELECT * FROM clientes")
todos_clientes = cursor.fetchall()
print("📋 Todos os clientes:")
for cliente in todos_clientes:
    print(f"  {cliente[0]} - {cliente[1]} - {cliente[4]}")

print("\n" + "="*50 + "\n")

# SELECT específico - Nome e cidade
cursor.execute("SELECT nome, cidade FROM clientes")
nomes_cidades = cursor.fetchall()
print("📋 Nome e cidade dos clientes:")
for cliente in nomes_cidades:
    print(f"  {cliente[0]} - {cliente[1]}")
```
