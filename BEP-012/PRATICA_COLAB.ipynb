## 1. Configuração Inicial

Primeiro, vamos importar o módulo sqlite3 e criar um banco de dados em memória.


```python
import sqlite3
import pandas as pd

# Criar conexão com banco em memória
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

print("✅ Conexão com SQLite estabelecida!")
print("📊 Banco de dados criado em memória")
```

## 2. Comandos DDL - CREATE, ALTER, DROP

### 2.1 Criando Tabelas (CREATE TABLE)


```python
# Criar tabela de clientes
cursor.execute('''
CREATE TABLE clientes (
    id INTEGER PRIMARY KEY,
    nome TEXT NOT NULL,
    email TEXT,
    telefone TEXT,
    cidade TEXT
)
''')

print("✅ Tabela 'clientes' criada com sucesso!")

# Criar tabela de produtos
cursor.execute('''
CREATE TABLE produtos (
    id INTEGER PRIMARY KEY,
    nome TEXT NOT NULL,
    preco REAL,
    estoque INTEGER
)
''')

print("✅ Tabela 'produtos' criada com sucesso!")
```

### 2.2 Modificando Tabelas (ALTER TABLE)


```python
# Adicionar coluna 'data_cadastro' na tabela clientes
cursor.execute('''
ALTER TABLE clientes ADD COLUMN data_cadastro DATE
''')

print("✅ Coluna 'data_cadastro' adicionada à tabela 'clientes'!")

# Verificar a estrutura da tabela
cursor.execute("PRAGMA table_info(clientes)")
colunas = cursor.fetchall()

print("\n📋 Estrutura da tabela 'clientes':")
for coluna in colunas:
    print(f"  - {coluna[1]} ({coluna[2]})")
```

## 3. Inserindo Dados (INSERT)


```python
# Inserir clientes
clientes = [
    ('João Silva', 'joao@email.com', '(71) 99999-1111', 'Salvador', '2024-01-15'),
    ('Maria Santos', 'maria@email.com', '(71) 99999-2222', 'Feira de Santana', '2024-01-20'),
    ('Pedro Costa', 'pedro@email.com', '(71) 99999-3333', 'Salvador', '2024-01-25'),
    ('Ana Lima', 'ana@email.com', '(71) 99999-4444', 'Vitória da Conquista', '2024-02-01'),
    ('Carlos Oliveira', 'carlos@email.com', '(71) 99999-5555', 'Camaçari', '2024-02-05')
]

cursor.executemany('''
INSERT INTO clientes (nome, email, telefone, cidade, data_cadastro)
VALUES (?, ?, ?, ?, ?)
''', clientes)

print(f"✅ {len(clientes)} clientes inseridos com sucesso!")
```

```python
# Inserir produtos
produtos = [
    ('Notebook Dell', 2500.00, 15),
    ('Mouse Logitech', 89.90, 50),
    ('Teclado Mecânico', 199.90, 25),
    ('Monitor 24"', 800.00, 10),
    ('Webcam HD', 150.00, 30)
]

cursor.executemany('''
INSERT INTO produtos (nome, preco, estoque)
VALUES (?, ?, ?)
''', produtos)

print(f"✅ {len(produtos)} produtos inseridos com sucesso!")
```

## 4. Consultando Dados (SELECT)

### 4.1 SELECT Básico


```python
# Selecionar todos os clientes
cursor.execute("SELECT * FROM clientes")
clientes = cursor.fetchall()

print("👥 Todos os clientes:")
for cliente in clientes:
    print(f"  ID: {cliente[0]}, Nome: {cliente[1]}, Email: {cliente[2]}, Cidade: {cliente[4]}")
```

```python
# Selecionar apenas nome e email dos clientes
cursor.execute("SELECT nome, email FROM clientes")
clientes = cursor.fetchall()

print("📧 Nome e email dos clientes:")
for cliente in clientes:
    print(f"  {cliente[0]} - {cliente[1]}")
```

### 4.2 Filtros (WHERE, ORDER BY, LIMIT)


```python
# Clientes de Salvador
cursor.execute("SELECT nome, email, cidade FROM clientes WHERE cidade = 'Salvador'")
clientes_salvador = cursor.fetchall()

print("🏙️ Clientes de Salvador:")
for cliente in clientes_salvador:
    print(f"  {cliente[0]} - {cliente[1]}")
```

```python
# Produtos ordenados por preço (maior para menor)
cursor.execute("SELECT nome, preco FROM produtos ORDER BY preco DESC")
produtos_ordenados = cursor.fetchall()

print("💰 Produtos ordenados por preço (maior para menor):")
for produto in produtos_ordenados:
    print(f"  {produto[0]} - R$ {produto[1]:.2f}")
```

```python
# Mostrar apenas os 3 primeiros clientes
cursor.execute("SELECT nome, cidade FROM clientes LIMIT 3")
primeiros_clientes = cursor.fetchall()

print("👆 Os 3 primeiros clientes:")
for cliente in primeiros_clientes:
    print(f"  {cliente[0]} - {cliente[1]}")
```

```python
# Combinando filtros: produtos com estoque > 20, ordenados por nome
cursor.execute("""
SELECT nome, preco, estoque 
FROM produtos 
WHERE estoque > 20 
ORDER BY nome
""")
produtos_filtrados = cursor.fetchall()

print("📦 Produtos com estoque > 20, ordenados por nome:")
for produto in produtos_filtrados:
    print(f"  {produto[0]} - R$ {produto[1]:.2f} (Estoque: {produto[2]})")
```

## 5. Exercícios Práticos

Agora é sua vez! Complete os exercícios abaixo:


### Exercício 1: Consultas Básicas

**Tarefa:** Escreva consultas SQL para:

1. Mostrar todos os produtos com preço menor que R$ 200,00
2. Listar clientes cujo nome começa com 'A'
3. Mostrar os 2 produtos mais caros


```python
# Sua solução aqui:

# 1. Produtos com preço < R$ 200,00
cursor.execute("SELECT nome, preco FROM produtos WHERE preco < 200")
produtos_baratos = cursor.fetchall()
print("1. Produtos com preço < R$ 200,00:")
for produto in produtos_baratos:
    print(f"   {produto[0]} - R$ {produto[1]:.2f}")

print()

# 2. Clientes com nome começando com 'A'
cursor.execute("SELECT nome, email FROM clientes WHERE nome LIKE 'A%'")
clientes_a = cursor.fetchall()
print("2. Clientes com nome começando com 'A':")
for cliente in clientes_a:
    print(f"   {cliente[0]} - {cliente[1]}")

print()

# 3. Os 2 produtos mais caros
cursor.execute("SELECT nome, preco FROM produtos ORDER BY preco DESC LIMIT 2")
produtos_caros = cursor.fetchall()
print("3. Os 2 produtos mais caros:")
for produto in produtos_caros:
    print(f"   {produto[0]} - R$ {produto[1]:.2f}")
```

### Exercício 2: Inserção de Dados

**Tarefa:** 

1. Insira um novo cliente: 'Lucia Ferreira', 'lucia@email.com', '(71) 99999-6666', 'Ilhéus', '2024-02-10'
2. Insira um novo produto: 'Tablet Samsung', 1200.00, 8
3. Verifique se os dados foram inseridos corretamente


```python
# Sua solução aqui:

# 1. Inserir novo cliente
cursor.execute('''
INSERT INTO clientes (nome, email, telefone, cidade, data_cadastro)
VALUES ('Lucia Ferreira', 'lucia@email.com', '(71) 99999-6666', 'Ilhéus', '2024-02-10')
''')
print("✅ Novo cliente inserido!")

# 2. Inserir novo produto
cursor.execute('''
INSERT INTO produtos (nome, preco, estoque)
VALUES ('Tablet Samsung', 1200.00, 8)
''')
print("✅ Novo produto inserido!")

# 3. Verificar os dados inseridos
print("\n👥 Verificando clientes:")
cursor.execute("SELECT nome, cidade FROM clientes WHERE nome = 'Lucia Ferreira'")
cliente_novo = cursor.fetchone()
if cliente_novo:
    print(f"   {cliente_novo[0]} - {cliente_novo[1]}")

print("\n📱 Verificando produtos:")
cursor.execute("SELECT nome, preco, estoque FROM produtos WHERE nome = 'Tablet Samsung'")
produto_novo = cursor.fetchone()
if produto_novo:
    print(f"   {produto_novo[0]} - R$ {produto_novo[1]:.2f} (Estoque: {produto_novo[2]})")
```

## 6. Desafio Final

**Tarefa:** Crie uma consulta que mostre:
- Nome do cliente
- Cidade
- Data de cadastro
- Apenas clientes cadastrados em 2024
- Ordenados por data de cadastro (mais recente primeiro)
- Limitado a 3 resultados


```python
# Sua solução aqui:

cursor.execute('''
SELECT nome, cidade, data_cadastro
FROM clientes
WHERE data_cadastro LIKE '2024%'
ORDER BY data_cadastro DESC
LIMIT 3
''')

clientes_2024 = cursor.fetchall()
print("🏆 Desafio: Clientes cadastrados em 2024 (mais recentes primeiro):")
for cliente in clientes_2024:
    print(f"   {cliente[0]} - {cliente[1]} - {cliente[2]}")
```
