```python
# Exemplo 1: Herança básica
class Animal:
    def __init__(self, nome):
        self.nome = nome
    
    def fazer_som(self):
        return "Fazendo algum som..."

class Cachorro(Animal):
    def latir(self):
        return f"{self.nome}: Au au!"

# Cachorro herda nome e fazer_som() de Animal
rex = Cachorro("Rex")
print(rex.nome)  # Herdado
print(rex.fazer_som())  # Herdado
print(rex.latir())  # Próprio de Cachorro

```

```python
# Exemplo 2: Usando super() no construtor
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade
    
    def apresentar(self):
        return f"Sou {self.nome}, tenho {self.idade} anos"

class Estudante(Pessoa):
    def __init__(self, nome, idade, matricula):
        super().__init__(nome, idade)  # Chama construtor da classe pai
        self.matricula = matricula

est = Estudante("Ana", 20, "2025A001")
print(est.nome)  # Herdado
print(est.idade)  # Herdado
print(est.matricula)  # Próprio
print(est.apresentar())  # Herdado

```

```python
# Exemplo 3: Sobrescrita de métodos (Override)
class Veiculo:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo
        self.velocidade = 0
    
    def acelerar(self):
        self.velocidade += 10
        return f"{self.modelo} acelerou para {self.velocidade} km/h"

class Carro(Veiculo):
    def acelerar(self):  # Sobrescreve o método
        self.velocidade += 15
        return f"{self.modelo} acelerou rapidamente para {self.velocidade} km/h"

class Moto(Veiculo):
    def acelerar(self):  # Sobrescreve o método
        self.velocidade += 20
        return f"{self.modelo} acelerou muito rápido para {self.velocidade} km/h"

carro = Carro("Toyota", "Corolla")
moto = Moto("Honda", "CB600")

print(carro.acelerar())
print(moto.acelerar())

```

```python
# Exemplo 4: Polimorfismo - Animal e subclasses
class Animal:
    def fazer_som(self):
        return "Fazendo algum som..."

class Cachorro(Animal):
    def fazer_som(self):
        return "Au au!"

class Gato(Animal):
    def fazer_som(self):
        return "Miau!"

class Vaca(Animal):
    def fazer_som(self):
        return "Muuu!"

# Polimorfismo: mesmo método, comportamentos diferentes
animais = [Cachorro(), Gato(), Vaca()]

for animal in animais:
    print(animal.fazer_som())

```

```python
# Exemplo 5: Polimorfismo com Formas
import math

class Forma:
    def calcular_area(self):
        return 0
    
    def calcular_perimetro(self):
        return 0

class Retangulo(Forma):
    def __init__(self, largura, altura):
        self.largura = largura
        self.altura = altura
    
    def calcular_area(self):
        return self.largura * self.altura
    
    def calcular_perimetro(self):
        return 2 * (self.largura + self.altura)

class Circulo(Forma):
    def __init__(self, raio):
        self.raio = raio
    
    def calcular_area(self):
        return math.pi * self.raio ** 2
    
    def calcular_perimetro(self):
        return 2 * math.pi * self.raio

# Polimorfismo: tratando todas as formas uniformemente
formas = [
    Retangulo(5, 3),
    Circulo(4),
    Retangulo(2, 2)
]

for forma in formas:
    print(f"Área: {forma.calcular_area():.2f}, Perímetro: {forma.calcular_perimetro():.2f}")

```

```python
# Exercício 1: Hierarquia de Funcionários
class Funcionario:
    def __init__(self, nome, salario_base):
        self.nome = nome
        self.salario_base = salario_base
    
    def calcular_salario(self):
        return self.salario_base

class Gerente(Funcionario):
    def calcular_salario(self):
        return self.salario_base * 1.20  # 20% de bônus

class Vendedor(Funcionario):
    def __init__(self, nome, salario_base):
        super().__init__(nome, salario_base)
        self.vendas = 0
    
    def adicionar_venda(self, valor):
        self.vendas += valor
    
    def calcular_salario(self):
        comissao = self.vendas * 0.10  # 10% de comissão
        return self.salario_base + comissao

gerente = Gerente("Maria", 5000)
vendedor = Vendedor("João", 2000)
vendedor.adicionar_venda(10000)

print(f"{gerente.nome}: R$ {gerente.calcular_salario():.2f}")
print(f"{vendedor.nome}: R$ {vendedor.calcular_salario():.2f}")

```

```python
# Exercício 2: Hierarquia de Contas Bancárias
class Conta:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.saldo = saldo_inicial
    
    def depositar(self, valor):
        if valor > 0:
            self.saldo += valor
    
    def sacar(self, valor):
        if valor > 0 and valor <= self.saldo:
            self.saldo -= valor
            return True
        return False

class ContaCorrente(Conta):
    def sacar(self, valor):
        taxa = valor * 0.02  # 2% de taxa
        valor_total = valor + taxa
        if valor_total <= self.saldo:
            self.saldo -= valor_total
            return True
        return False

class ContaPoupanca(Conta):
    def aplicar_juros(self, taxa=0.5):
        juros = self.saldo * (taxa / 100)
        self.saldo += juros
        return juros

cc = ContaCorrente("Ana", 1000)
cp = ContaPoupanca("Bruno", 1000)

cc.sacar(100)
cp.aplicar_juros()

print(f"Conta Corrente: R$ {cc.saldo:.2f}")
print(f"Conta Poupança: R$ {cp.saldo:.2f}")

```

```python
# Exercício 3: Sistema de Notificações (Polimorfismo)
class Notificacao:
    def __init__(self, destinatario, mensagem):
        self.destinatario = destinatario
        self.mensagem = mensagem
    
    def enviar(self):
        return "Enviando notificação..."

class Email(Notificacao):
    def enviar(self):
        return f"Email enviado para {self.destinatario}: {self.mensagem}"

class SMS(Notificacao):
    def enviar(self):
        return f"SMS enviado para {self.destinatario}: {self.mensagem}"

class Push(Notificacao):
    def enviar(self):
        return f"Push notification enviada para {self.destinatario}: {self.mensagem}"

# Polimorfismo: enviando todas de forma uniforme
notificacoes = [
    Email("ana@email.com", "Olá!"),
    SMS("11999999999", "Olá!"),
    Push("user123", "Olá!")
]

for notif in notificacoes:
    print(notif.enviar())

```

```python
# Exercício 4: Hierarquia de Produtos com Polimorfismo
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco
    
    def calcular_desconto(self):
        return 0
    
    def preco_final(self):
        desconto = self.calcular_desconto()
        return self.preco - desconto

class ProdutoEletronico(Produto):
    def calcular_desconto(self):
        return self.preco * 0.10

class ProdutoAlimenticio(Produto):
    def calcular_desconto(self):
        return self.preco * 0.05

class ProdutoVestuario(Produto):
    def calcular_desconto(self):
        return self.preco * 0.15

# Polimorfismo: calculando descontos de forma uniforme
produtos = [
    ProdutoEletronico("Notebook", 3000),
    ProdutoAlimenticio("Arroz", 20),
    ProdutoVestuario("Camiseta", 50)
]

for produto in produtos:
    print(f"{produto.nome}: R$ {produto.preco:.2f} -> R$ {produto.preco_final():.2f}")

```

```python
# Exemplo Completo: Sistema de Veículos
class Veiculo:
    def __init__(self, marca, modelo, ano):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
        self.velocidade = 0
        self.ligado = False
    
    def ligar(self):
        self.ligado = True
        return f"{self.modelo} ligado"
    
    def acelerar(self, incremento=10):
        if self.ligado:
            self.velocidade += incremento
            return f"{self.modelo} acelerou para {self.velocidade} km/h"
        return f"{self.modelo} precisa estar ligado"
    
    def informacoes(self):
        return f"{self.marca} {self.modelo} ({self.ano}) - {self.velocidade} km/h"

class Carro(Veiculo):
    def __init__(self, marca, modelo, ano, portas):
        super().__init__(marca, modelo, ano)
        self.portas = portas
    
    def acelerar(self, incremento=15):
        if self.ligado:
            self.velocidade += incremento
            return f"{self.modelo} acelerou suavemente para {self.velocidade} km/h"
        return f"{self.modelo} precisa estar ligado"

class Moto(Veiculo):
    def __init__(self, marca, modelo, ano, cilindradas):
        super().__init__(marca, modelo, ano)
        self.cilindradas = cilindradas
    
    def acelerar(self, incremento=25):
        if self.ligado:
            self.velocidade += incremento
            return f"{self.modelo} acelerou rapidamente para {self.velocidade} km/h"
        return f"{self.modelo} precisa estar ligado"

# Demonstração de polimorfismo
veiculos = [
    Carro("Toyota", "Corolla", 2023, 4),
    Moto("Honda", "CB600", 2022, 600)
]

for veiculo in veiculos:
    print(veiculo.ligar())
    print(veiculo.acelerar())
    print(veiculo.informacoes())
    print()

```
