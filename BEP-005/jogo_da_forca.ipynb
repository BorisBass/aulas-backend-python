# Projeto Final: Jogo da Forca em Python

Este notebook contém o código para o Jogo da Forca, um projeto prático que utiliza os conceitos de:

*   **Estruturas de Repetição** (`while`, `for`)
*   **Estruturas Condicionais** (`if`, `elif`, `else`)
*   **Variáveis, Tipos de Dados e Entrada/Saída**

### Como Jogar:
1.  Execute a célula de código abaixo.
2.  Siga as instruções no console para chutar letras e tentar adivinhar a palavra secreta.
3.  Divirta-se e pratique seus conhecimentos em Python!

```

import random

def jogar_forca():
    palavras = ["python", "programacao", "computador", "desenvolvimento", "algoritmo", "inteligencia"]
    palavra_secreta = random.choice(palavras).upper()
    letras_descobertas = ["_" for _ in palavra_secreta]
    letras_erradas = []
    tentativas = 6

    print("Bem-vindo ao Jogo da Forca!")
    print("Adivinhe a palavra secreta.")

    while tentativas > 0 and "_" in letras_descobertas:
        print("
" + "-"*30)
        print(f"Palavra: {" ".join(letras_descobertas)}")
        print(f"Tentativas restantes: {tentativas}")
        print(f"Letras erradas: {", ".join(letras_erradas)}")

        chute = input("Chute uma letra: ").upper()

        if len(chute) != 1 or not chute.isalpha():
            print("Por favor, digite apenas uma letra válida.")
            continue

        if chute in letras_descobertas or chute in letras_erradas:
            print(f"Você já chutou a letra '{chute}'. Tente outra.")
            continue

        if chute in palavra_secreta:
            print(f"Boa! A letra '{chute}' está na palavra.")
            for i in range(len(palavra_secreta)):
                if palavra_secreta[i] == chute:
                    letras_descobertas[i] = chute
        else:
            print(f"Que pena! A letra '{chute}' não está na palavra.")
            letras_erradas.append(chute)
            tentativas -= 1

    print("
" + "="*30)
    if "_" not in letras_descobertas:
        print(f"Parabéns! Você adivinhou a palavra: {palavra_secreta}")
        print("Você VENCEU!")
    else:
        print(f"Suas tentativas acabaram! A palavra era: {palavra_secreta}")
        print("Você PERDEU!")
    print("="*30)

# Para rodar o jogo, chame a função:
jogar_forca()
```
