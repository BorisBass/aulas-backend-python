```python
# Abordagem procedural
nome = "Caderno"
preco = 20.00

def aplicar_desconto(preco, percentual):
    fator = 1 - (percentual / 100)
    return round(preco * fator, 2)

novo_preco = aplicar_desconto(preco, 10)
print(nome, novo_preco)
```

```python
# Abordagem orientada a objetos
class Animal:
    def __init__(self, nome, especie):
        self.nome = nome
        self.especie = especie
    
    def fazer_som(self):
        return f"{self.nome} faz um som!"

gato = Animal("Mimi", "Gato")
print(gato.fazer_som())
```

```python
# Classe e objeto passo a passo
class Aluno:
    pass

a = Aluno()
print(type(a))  # <class '__main__.Aluno'>
```

```python
# Exemplo: Classe Veiculo
class Veiculo:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo
        self.velocidade = 0  # Atributo inicializado
    
    def acelerar(self):
        self.velocidade += 10
        return f"{self.modelo} acelerou! Velocidade: {self.velocidade} km/h"

carro = Veiculo("Toyota", "Corolla")
print(carro.acelerar())
print(carro.acelerar())
```

```python
# Exemplo: Classe Livro (conceitual)
class Livro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor
        self.paginas_lidas = 0
    
    def ler_paginas(self, quantidade):
        self.paginas_lidas += quantidade
        return f"Você leu {quantidade} páginas de '{self.titulo}'"

livro = Livro("Python para Iniciantes", "João Silva")
print(livro.ler_paginas(10))
print(livro.ler_paginas(5))
```

```python
# Exercício guiado: Lampada
class Lampada:
    def __init__(self, cor):
        self.cor = cor
        self.ligada = False
    
    def ligar(self):
        self.ligada = True
        return f"Lâmpada {self.cor} ligada!"
    
    def desligar(self):
        self.ligada = False
        return f"Lâmpada {self.cor} desligada!"
    
    def status(self):
        return "ligada" if self.ligada else "desligada"

lamp = Lampada("branca")
print(lamp.ligar())
print(f"Status: {lamp.status()}")
```

```python
# Exercício guiado: Contador
class Contador:
    def __init__(self, nome):
        self.nome = nome
        self.valor = 0
    
    def incrementar(self):
        self.valor += 1
        return f"{self.nome}: {self.valor}"
    
    def zerar(self):
        self.valor = 0
        return f"{self.nome} zerado!"

cont = Contador("Contador de cliques")
print(cont.incrementar())
print(cont.incrementar())
print(cont.zerar())
```

```python
# Mini-projeto: ContaBancaria
class ContaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.saldo = float(saldo_inicial)

    def depositar(self, valor):
        if valor > 0:
            self.saldo += valor

    def sacar(self, valor):
        if 0 <= valor <= self.saldo:
            self.saldo -= valor

    def ver_saldo(self):
        return round(self.saldo, 2)

c = ContaBancaria("Ana")
c.depositar(100)
c.sacar(30)
print(c.titular, c.ver_saldo())
```
