```python
# Abordagem procedural
nome = "Caderno"
preco = 20.00

def aplicar_desconto(preco, percentual):
    fator = 1 - (percentual / 100)
    return round(preco * fator, 2)

novo_preco = aplicar_desconto(preco, 10)
print(nome, novo_preco)
```

```python
# Abordagem orientada a objetos
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = float(preco)

    def aplicar_desconto(self, percentual):
        fator = 1 - (percentual / 100)
        self.preco = round(self.preco * fator, 2)

p = Produto("Caderno", 20)
p.aplicar_desconto(10)
print(p.nome, p.preco)
```

```python
# Classe e objeto passo a passo
class Aluno:
    pass

a = Aluno()
print(type(a))  # <class '__main__.Aluno'>
```

```python
# Adicionando atributos com __init__
class Aluno:
    def __init__(self, nome, matricula):
        self.nome = nome
        self.matricula = matricula

al = Aluno("Ana", "2025A001")
print(al.nome)
print(al.matricula)
```

```python
# Métodos e self
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = float(preco)

    def aplicar_desconto(self, percentual):
        fator = 1 - (percentual / 100)
        self.preco = round(self.preco * fator, 2)

p = Produto("Caderno", 20)
p.aplicar_desconto(10)
print(p.preco)
```

```python
# Exercício guiado: Aluno.aprovado
class Aluno:
    def __init__(self, nome, nota):
        self.nome = nome
        self.nota = float(nota)

    def aprovado(self):
        return self.nota >= 6

al1 = Aluno("Ana", 7.5)
al2 = Aluno("Bruno", 5.8)
print(al1.nome, al1.aprovado())
print(al2.nome, al2.aprovado())
```

```python
# Exercício guiado: Produto.reajustar
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = float(preco)

    def reajustar(self, percentual):
        fator = 1 + (percentual / 100)
        self.preco = round(self.preco * fator, 2)

p = Produto("Caderno", 20)
p.reajustar(10)
print(p.preco)
```

```python
# Mini-projeto: ContaBancaria
class ContaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.saldo = float(saldo_inicial)

    def depositar(self, valor):
        if valor > 0:
            self.saldo += valor

    def sacar(self, valor):
        if 0 <= valor <= self.saldo:
            self.saldo -= valor

    def ver_saldo(self):
        return round(self.saldo, 2)

c = ContaBancaria("Ana")
c.depositar(100)
c.sacar(30)
print(c.titular, c.ver_saldo())
```
