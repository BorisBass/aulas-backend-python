```python
# Problema sem encapsulamento
class ContaBancaria:
    def __init__(self, titular, saldo):
        self.titular = titular
        self.saldo = saldo  # Acesso direto - perigoso!

conta = ContaBancaria("Ana", 1000)
print(f"Saldo inicial: R$ {conta.saldo:.2f}")

# Problema: pode modificar diretamente
conta.saldo = -500  # ❌ Saldo negativo!
conta.saldo = "erro"  # ❌ Tipo errado!
print(f"Saldo após modificação indevida: {conta.saldo}")

```

```python
# Modificadores de acesso
class Exemplo:
    def __init__(self):
        self.publico = "Público"  # Acesso de qualquer lugar
        self._protegido = "Protegido"  # Convenção (não acessar diretamente)
        self.__privado = "Privado"  # Acesso apenas na classe

obj = Exemplo()
print(obj.publico)  # ✅ OK
print(obj._protegido)  # ⚠️ Funciona, mas não recomendado
# print(obj.__privado)  # ❌ Erro! AttributeError

```

```python
# Encapsulamento: ContaBancaria com saldo privado (usando @property)
class ContaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.__saldo = float(saldo_inicial)  # Privado
    
    @property
    def saldo(self):
        """Getter: acessa saldo como atributo"""
        return self.__saldo
    
    # Métodos de instância para manipular saldo
    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor
            return f"Depósito de R$ {valor:.2f} realizado."
        return "Valor inválido para depósito."
    
    def sacar(self, valor):
        if valor > 0 and valor <= self.__saldo:
            self.__saldo -= valor
            return f"Saque de R$ {valor:.2f} realizado."
        return "Saldo insuficiente ou valor inválido."

conta = ContaBancaria("Ana", 1000)
print(f"Saldo: R$ {conta.saldo:.2f}")  # ✅ Acesso elegante como atributo
print(conta.depositar(500))
print(conta.sacar(200))
print(f"Saldo final: R$ {conta.saldo:.2f}")

```

```python
# Getters e Setters Tradicionais (Forma Antiga - apenas para referência)
# ⚠️ NOTA: Esta é a forma tradicional. Veja a célula seguinte para a forma pythônica com @property
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.__preco = 0
        self.set_preco(preco)  # Usa setter na inicialização
    
    # Getter
    def get_preco(self):
        return self.__preco
    
    # Setter com validação
    def set_preco(self, novo_preco):
        if isinstance(novo_preco, (int, float)) and novo_preco > 0:
            self.__preco = float(novo_preco)
        else:
            raise ValueError("Preço deve ser um número positivo")

p = Produto("Notebook", 2500)
print(f"Preço: R$ {p.get_preco():.2f}")  # Forma tradicional
p.set_preco(2200)
print(f"Novo preço: R$ {p.get_preco():.2f}")

```

```python
# @property: Forma Pythônica e Elegante
class ProdutoElegante:
    def __init__(self, nome, preco):
        self.nome = nome
        self.__preco = 0
        self.preco = preco  # Usa o setter na inicialização
    
    @property
    def preco(self):
        """Getter: acessa como atributo"""
        return self.__preco
    
    @preco.setter
    def preco(self, valor):
        """Setter: valida antes de atribuir"""
        if isinstance(valor, (int, float)) and valor > 0:
            self.__preco = float(valor)
        else:
            raise ValueError("Preço deve ser um número positivo")

# Uso elegante: parece atributo, mas tem validação!
p = ProdutoElegante("Notebook", 2500)
print(f"Preço: R$ {p.preco:.2f}")  # Acesso direto como atributo
p.preco = 2200  # Atribuição direta como atributo
print(f"Novo preço: R$ {p.preco:.2f}")

# Comparação:
# Tradicional: p.get_preco() e p.set_preco(2200)
# Pythônico:   p.preco e p.preco = 2200

```

```python
# Exemplo completo com @property: Pessoa
class PessoaElegante:
    def __init__(self, nome, idade):
        self.__nome = ""
        self.__idade = 0
        self.nome = nome  # Usa setter
        self.idade = idade  # Usa setter
    
    @property
    def nome(self):
        return self.__nome
    
    @nome.setter
    def nome(self, valor):
        if isinstance(valor, str) and len(valor) > 0:
            self.__nome = valor
        else:
            raise ValueError("Nome deve ser uma string não vazia")
    
    @property
    def idade(self):
        return self.__idade
    
    @idade.setter
    def idade(self, valor):
        if isinstance(valor, int) and valor >= 0:
            self.__idade = valor
        else:
            raise ValueError("Idade deve ser um inteiro não negativo")
    
    def fazer_aniversario(self):
        self.idade += 1  # Usa o setter automaticamente
        return f"{self.nome} fez aniversário! Agora tem {self.idade} anos."

p = PessoaElegante("Ana", 25)
print(f"{p.nome} tem {p.idade} anos")
p.idade = 26  # Atribuição direta com validação
print(p.fazer_aniversario())

```

```python
# Exercício 1: Aluno Encapsulada (com @property)
class Aluno:
    def __init__(self, nome, nota):
        self.nome = nome
        self.__nota = 0
        self.nota = nota  # Usa o setter para validar
    
    @property
    def nota(self):
        return self.__nota
    
    @nota.setter
    def nota(self, valor):
        if 0 <= valor <= 10:
            self.__nota = valor
        else:
            raise ValueError("Nota deve estar entre 0 e 10")
    
    def aprovado(self):
        return self.__nota >= 6

a = Aluno("Ana", 7.5)
print(f"{a.nome}: Nota {a.nota} - {'Aprovado' if a.aprovado() else 'Reprovado'}")  # Sintaxe elegante
a.nota = 8.0  # Atribuição direta com validação
print(f"Nova nota: {a.nota}")

```

```python
# Exercício 2: Produto com múltiplos atributos privados (com @property)
class Produto:
    def __init__(self, nome, preco, estoque):
        self.nome = nome
        self.__preco = 0
        self.__estoque = 0
        self.preco = preco  # Usa o setter
        self.estoque = estoque  # Usa o setter
    
    @property
    def preco(self):
        return self.__preco
    
    @preco.setter
    def preco(self, valor):
        if valor > 0:
            self.__preco = float(valor)
        else:
            raise ValueError("Preço deve ser maior que zero")
    
    @property
    def estoque(self):
        return self.__estoque
    
    @estoque.setter
    def estoque(self, valor):
        if valor >= 0:
            self.__estoque = int(valor)
        else:
            raise ValueError("Estoque não pode ser negativo")
    
    def vender(self, quantidade):
        if quantidade > 0 and quantidade <= self.__estoque:
            self.__estoque -= quantidade
            return f"Venda de {quantidade} unidade(s) realizada. Estoque: {self.__estoque}"
        return "Estoque insuficiente ou quantidade inválida."

p = Produto("Notebook", 2500, 10)
print(f"Preço: R$ {p.preco:.2f}, Estoque: {p.estoque}")  # Sintaxe elegante
print(p.vender(3))
print(f"Estoque após venda: {p.estoque}")

```

```python
# Exercício 3: Pessoa com idade (com @property)
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.__idade = 0  # Private
        self.idade = idade  # Usa o setter
    
    @property
    def idade(self):
        return self.__idade
    
    @idade.setter
    def idade(self, valor):
        if valor >= 0:
            self.__idade = int(valor)
        else:
            raise ValueError("Idade não pode ser negativa")
    
    def fazer_aniversario(self):
        self.idade += 1  # Usa o setter automaticamente
        return f"{self.nome} fez aniversário! Agora tem {self.idade} anos."

p = Pessoa("Ana", 25)
print(f"Idade: {p.idade} anos")  # Sintaxe elegante
print(p.fazer_aniversario())

```

```python
# Exemplo completo: ContaBancaria com transferência (usando @property)
class ContaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.__saldo = float(saldo_inicial)
    
    @property
    def saldo(self):
        """Getter: acessa saldo como atributo"""
        return self.__saldo
    
    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor
            return f"Depósito de R$ {valor:.2f} realizado."
        return "Valor inválido para depósito."
    
    def sacar(self, valor):
        if valor > 0 and valor <= self.__saldo:
            self.__saldo -= valor
            return f"Saque de R$ {valor:.2f} realizado."
        return "Saldo insuficiente ou valor inválido."
    
    def transferir(self, outra_conta, valor):
        if self.sacar(valor) != "Saldo insuficiente ou valor inválido.":
            outra_conta.depositar(valor)
            return f"Transferência de R$ {valor:.2f} realizada."
        return "Transferência não realizada."

conta1 = ContaBancaria("Ana", 1000)
conta2 = ContaBancaria("Bruno", 500)

print(f"Ana: R$ {conta1.saldo:.2f}")  # Sintaxe elegante
print(f"Bruno: R$ {conta2.saldo:.2f}")
conta1.transferir(conta2, 150)
print(f"Após transferência:")
print(f"Ana: R$ {conta1.saldo:.2f}")
print(f"Bruno: R$ {conta2.saldo:.2f}")

```
