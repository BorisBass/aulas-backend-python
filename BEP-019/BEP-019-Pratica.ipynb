```python
# Problema sem encapsulamento
class ContaBancaria:
    def __init__(self, titular, saldo):
        self.titular = titular
        self.saldo = saldo  # Acesso direto - perigoso!

conta = ContaBancaria("Ana", 1000)
print(f"Saldo inicial: R$ {conta.saldo:.2f}")

# Problema: pode modificar diretamente
conta.saldo = -500  # ❌ Saldo negativo!
conta.saldo = "erro"  # ❌ Tipo errado!
print(f"Saldo após modificação indevida: {conta.saldo}")

```

```python
# Modificadores de acesso
class Exemplo:
    def __init__(self):
        self.publico = "Público"  # Acesso de qualquer lugar
        self._protegido = "Protegido"  # Convenção (não acessar diretamente)
        self.__privado = "Privado"  # Acesso apenas na classe

obj = Exemplo()
print(obj.publico)  # ✅ OK
print(obj._protegido)  # ⚠️ Funciona, mas não recomendado
# print(obj.__privado)  # ❌ Erro! AttributeError

```

```python
# Encapsulamento: ContaBancaria com saldo privado
class ContaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.__saldo = float(saldo_inicial)  # Privado
    
    # Getter - método para ler o saldo
    def get_saldo(self):
        return self.__saldo
    
    # Métodos de instância para manipular saldo
    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor
            return f"Depósito de R$ {valor:.2f} realizado."
        return "Valor inválido para depósito."
    
    def sacar(self, valor):
        if valor > 0 and valor <= self.__saldo:
            self.__saldo -= valor
            return f"Saque de R$ {valor:.2f} realizado."
        return "Saldo insuficiente ou valor inválido."

conta = ContaBancaria("Ana", 1000)
print(conta.get_saldo())  # ✅ Via método
print(conta.depositar(500))
print(conta.sacar(200))
print(f"Saldo final: R$ {conta.get_saldo():.2f}")

```

```python
# Getters e Setters
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.__preco = 0
        self.set_preco(preco)  # Usa setter na inicialização
    
    # Getter
    def get_preco(self):
        return self.__preco
    
    # Setter com validação
    def set_preco(self, novo_preco):
        if isinstance(novo_preco, (int, float)) and novo_preco > 0:
            self.__preco = float(novo_preco)
        else:
            raise ValueError("Preço deve ser um número positivo")

p = Produto("Notebook", 2500)
print(f"Preço: R$ {p.get_preco():.2f}")
p.set_preco(2200)
print(f"Novo preço: R$ {p.get_preco():.2f}")

```

```python
# Exercício 1: Aluno Encapsulada
class Aluno:
    def __init__(self, nome, nota):
        self.nome = nome
        self.__nota = 0
        self.set_nota(nota)  # Usa setter para validar
    
    def get_nota(self):
        return self.__nota
    
    def set_nota(self, nota):
        if 0 <= nota <= 10:
            self.__nota = nota
        else:
            raise ValueError("Nota deve estar entre 0 e 10")
    
    def aprovado(self):
        return self.__nota >= 6

a = Aluno("Ana", 7.5)
print(f"{a.nome}: Nota {a.get_nota()} - {'Aprovado' if a.aprovado() else 'Reprovado'}")

```

```python
# Exercício 2: Produto com múltiplos atributos privados
class Produto:
    def __init__(self, nome, preco, estoque):
        self.nome = nome
        self.__preco = 0
        self.__estoque = 0
        self.set_preco(preco)
        self.set_estoque(estoque)
    
    def get_preco(self):
        return self.__preco
    
    def set_preco(self, preco):
        if preco > 0:
            self.__preco = float(preco)
        else:
            raise ValueError("Preço deve ser maior que zero")
    
    def get_estoque(self):
        return self.__estoque
    
    def set_estoque(self, estoque):
        if estoque >= 0:
            self.__estoque = int(estoque)
        else:
            raise ValueError("Estoque não pode ser negativo")
    
    def vender(self, quantidade):
        if quantidade > 0 and quantidade <= self.__estoque:
            self.__estoque -= quantidade
            return f"Venda de {quantidade} unidade(s) realizada. Estoque: {self.__estoque}"
        return "Estoque insuficiente ou quantidade inválida."

p = Produto("Notebook", 2500, 10)
print(f"Preço: R$ {p.get_preco():.2f}, Estoque: {p.get_estoque()}")
print(p.vender(3))
print(f"Estoque após venda: {p.get_estoque()}")

```

```python
# Exercício 3: Pessoa com idade protegida
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self._idade = 0  # Protected (convenção)
        self.set_idade(idade)
    
    def get_idade(self):
        return self._idade
    
    def set_idade(self, idade):
        if idade >= 0:
            self._idade = int(idade)
        else:
            raise ValueError("Idade não pode ser negativa")
    
    def fazer_aniversario(self):
        self._idade += 1
        return f"{self.nome} fez aniversário! Agora tem {self._idade} anos."

p = Pessoa("Ana", 25)
print(f"Idade: {p.get_idade()} anos")
print(p.fazer_aniversario())

```

```python
# Exemplo completo: ContaBancaria com transferência
class ContaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.__saldo = float(saldo_inicial)
    
    def get_saldo(self):
        return self.__saldo
    
    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor
            return f"Depósito de R$ {valor:.2f} realizado."
        return "Valor inválido para depósito."
    
    def sacar(self, valor):
        if valor > 0 and valor <= self.__saldo:
            self.__saldo -= valor
            return f"Saque de R$ {valor:.2f} realizado."
        return "Saldo insuficiente ou valor inválido."
    
    def transferir(self, outra_conta, valor):
        if self.sacar(valor) != "Saldo insuficiente ou valor inválido.":
            outra_conta.depositar(valor)
            return f"Transferência de R$ {valor:.2f} realizada."
        return "Transferência não realizada."

conta1 = ContaBancaria("Ana", 1000)
conta2 = ContaBancaria("Bruno", 500)

print(f"Ana: R$ {conta1.get_saldo():.2f}")
print(f"Bruno: R$ {conta2.get_saldo():.2f}")
conta1.transferir(conta2, 150)
print(f"Após transferência:")
print(f"Ana: R$ {conta1.get_saldo():.2f}")
print(f"Bruno: R$ {conta2.get_saldo():.2f}")

```
