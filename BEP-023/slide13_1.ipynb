```python
# ❌ Sem Decorator - explosão de classes
class Cafe:
    def custo(self):
        return 5.0

class CafeComLeite(Cafe):
    def custo(self):
        return super().custo() + 2.0

class CafeComAcucar(Cafe):
    def custo(self):
        return super().custo() + 1.0

class CafeComLeiteEAcucar(Cafe):
    def custo(self):
        return super().custo() + 2.0 + 1.0

# E se quiser café com leite, açúcar E chantilly?
# Precisa criar outra classe!
# E se quiser outras combinações? Mais classes!

cafecomleite = CafeComLeite()
print(cafecomleite.custo)
```

```python
# ✅ Componente base
class Bebida:
    def __init__(self):
        self._descricao = "Bebida"
    
    def descricao(self):
        return self._descricao
    
    def custo(self):
        return 0.0

# Componente concreto
class Cafe(Bebida):
    def __init__(self):
        super().__init__()
        self._descricao = "Café"
    
    def custo(self):
        return 5.0

# Decorator abstrato
class Adicional(Bebida):
    def __init__(self, bebida):
        self._bebida = bebida
    
    def descricao(self):
        return self._bebida.descricao()
    
    def custo(self):
        return self._bebida.custo()

# Decorators concretos
class Leite(Adicional):
    def descricao(self):
        return self._bebida.descricao() + ", Leite"
    
    def custo(self):
        return self._bebida.custo() + 2.0

class Acucar(Adicional):
    def descricao(self):
        return self._bebida.descricao() + ", Açúcar"
    
    def custo(self):
        return self._bebida.custo() + 1.0

class Chantilly(Adicional):
    def descricao(self):
        return self._bebida.descricao() + ", Chantilly"
    
    def custo(self):
        return self._bebida.custo() + 3.0

class Canela(Adicional):
    def descricao(self):
        return self._bebida.descricao() + ", Mounjaro"
    
    def custo(self):
        return self._bebida.custo() + 10.0
```

```python

# Uso - composição dinâmica!
cafe = Cafe()
print(f"{cafe.descricao()}: R$ {cafe.custo():.2f}")
# Café: R$ 5.00

cafe_com_leite = Leite(Cafe())
print(f"{cafe_com_leite.descricao()}: R$ {cafe_com_leite.custo():.2f}")
# Café, Leite: R$ 7.00

cafe_completo = Chantilly(Acucar(Leite(Cafe())))
print(f"{cafe_completo.descricao()}: R$ {cafe_completo.custo():.2f}")
# Café, Leite, Açúcar, Chantilly: R$ 11.00
```

```python
cafe_mounjaro = Acucar(Canela(Cafe()))
print(f"{cafe_mounjaro.descricao()}: R$ {cafe_mounjaro.custo():.2f}")
```
