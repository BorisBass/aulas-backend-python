```python
class LoggerSimples:
    def __init__(self):
        self._logs = []
    
    def log(self, mensagem):
        self._logs.append(mensagem)
        print(f"[LOG] {mensagem}")
    
    def obter_logs(self):
        return self._logs.copy()

# Uso - CADA parte do sistema cria seu próprio logger
logger_a = LoggerSimples()
logger_a.log("Mensagem A")

logger_b = LoggerSimples()
logger_b.log("Mensagem B")

print(logger_a is logger_b)      # False - instâncias diferentes
print(logger_a)
print(logger_b)
print(logger_a.obter_logs())     # Só tem a mensagem A
print(logger_b.obter_logs())     # Só tem a mensagem B
```

```python
class Logger:
    _instancia = None
    
    def __init__(self):
        self._logs = []
    
    @classmethod
    def obter_instancia(cls):
        """Método que garante uma única instância"""
        if cls._instancia is None:
            cls._instancia = cls()  # Cria apenas na primeira vez logger1 = Logger()
        return cls._instancia
    
    def log(self, mensagem):
        self._logs.append(mensagem)
        print(f"[LOG] {mensagem}")
    
    def obter_logs(self):
        return self._logs.copy()

# Uso - IMPORTANTE: sempre usar obter_instancia()
loggerx = Logger()
loggerx.log("Mensagem pirata")

logger1 = Logger.obter_instancia()
logger1.log("Primeira mensagem")

logger2 = Logger.obter_instancia()  # Mesma instância!
logger2.log("Segunda mensagem")

print(logger1 is logger2)  # True
print(logger1)
print(logger2)
print(loggerx)
print(logger1.obter_logs())
print(logger2.obter_logs())
print(loggerx.obter_logs())  # Ambas as mensagens
```

```python
class Logger:
    _instancia = None
    
    def __new__(cls):
        """Controla a criação do objeto"""
        if cls._instancia is None:
            cls._instancia = super().__new__(cls)
        return cls._instancia
    
    def __init__(self):
        if not hasattr(self, 'inicializado'):
            self.inicializado = True
            self._logs = []
    
    def log(self, mensagem):
        self._logs.append(mensagem)
        print(f"[LOG] {mensagem}")
    
    def obter_logs(self):
        return self._logs.copy()

# Uso - Parece uma classe normal!
logger1 = Logger()  # Pode usar Logger() diretamente
logger1.log("Primeira mensagem")

logger2 = Logger()  # Mesma instância!
logger2.log("Segunda mensagem")

print(logger1 is logger2)  # True
print(logger1)
print(logger2)
print(logger1.obter_logs())
print(logger2.obter_logs())  # Ambas as mensagens
```
