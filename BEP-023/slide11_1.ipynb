```python
def notificar_clientes(produto, antigo, novo):
    print(f"🎉 Clientes: {produto} mudou de preço!")

def atualizar_estoque(produto, antigo, novo):
    print(f"📊 Estoque: atualizando preço de {produto}")

def atualiza_contabilidade(produto, antigo, novo):
    print(f"Contabilidade: produto {produto} mudou de {antigo} para {novo}")

class ProdutoSimples:
    def __init__(self, nome, preco):
        self.nome = nome
        self._preco = preco
    
    @property
    def preco(self):
        return self._preco
    
    @preco.setter
    def preco(self, novo_preco):
        if novo_preco != self._preco:
            antigo = self._preco
            self._preco = novo_preco
            # Chamadas diretas (acoplamento forte)
            notificar_clientes(self.nome, antigo, novo_preco)
            atualizar_estoque(self.nome, antigo, novo_preco)
            atualiza_contabilidade(self.nome, antigo, novo_preco)
```

```python
produto1 = ProdutoSimples("Mouse sem fio",80)
```

```python
print(f"{produto1.nome} - R$ {produto1.preco}")
produto1.preco = 90
print(f"{produto1.nome} - R$ {produto1.preco}")
```

```python
# Subject (Observado)
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self._preco = preco
        self._observers = []
    
    def adicionar_observer(self, observer):
        self._observers.append(observer)
    
    def remover_observer(self, observer):
        self._observers.remove(observer)
    
    @property
    def preco(self):
        return self._preco
    
    @preco.setter
    def preco(self, novo_preco):
        if novo_preco != self._preco:
            antigo_preco = self._preco
            self._preco = novo_preco
            self._notificar_mudanca_preco(antigo_preco, novo_preco)
    
    def _notificar_mudanca_preco(self, antigo, novo):
        for observer in self._observers:
            observer.preco_mudou(self, antigo, novo)

    def lista_observers(self):
        for observer in self._observers:
            print(f"{observer} ")
```

```python
# Observers

class ClienteObserver:
    def __init__(self, nome):
        self.nome = nome
    
    def preco_mudou(self, produto, antigo, novo):
        if novo < antigo:
            print(f"🎉 {self.nome}: {produto.nome} está mais barato!")
            print(f"   De R$ {antigo:.2f} para R$ {novo:.2f}")
        else:
            print(f"📢 {self.nome}: {produto.nome} aumentou de preço")
            print(f"   De R$ {antigo:.2f} para R$ {novo:.2f}")

class EstoqueObserver:
    def preco_mudou(self, produto, antigo, novo):
        print(f"📊 [ESTOQUE] Preço de {produto.nome} atualizado no sistema")
```

```python
class ContabilidadeObserver:
    def __init__(self, nome):
        self.nome = nome
            
    def preco_mudou(self, produto, antigo, novo):
        print(f"📊 [CONTABILIDADE] Preço de {produto.nome} atualizado no sistema, resolva a questão do imposto IBS")
```

```python
class EtiquetaDeGondola:
    def __init__(self, nome):
        self.nome = nome
            
    def preco_mudou(self, produto, antigo, novo):
        if novo < antigo:
            print(f"Emitir etiqueta de promoção!")
            print(f"🎉 {self.nome}: {produto.nome} PROMOÇÃO!")
            print(f"   De R$ {antigo:.2f} para R$ {novo:.2f}")
        else:
            print(f"Emitir etiqueta de alteração de preço.")
            print(f"📢 {self.nome}: {produto.nome}")
            print(f"   De R$ {antigo:.2f} para R$ {novo:.2f}")
```
