```python
class EmailNotificacao:
    def enviar(self, mensagem):
        print(f"📧 Email enviado: {mensagem}")

class SMSNotificacao:
    def enviar(self, mensagem):
        print(f"📱 SMS enviado: {mensagem}")

class PushNotificacao:
    def enviar(self, mensagem):
        print(f"🔔 Push enviado: {mensagem}")

class WppNotificacao:
    def enviar(self, mensagem):
        print(f"🔔 Wpp enviado: {mensagem}")

def enviar_notificacao(tipo, mensagem):
    if tipo == "email":
        notif = EmailNotificacao()
        notif.enviar(mensagem)
    elif tipo == "sms":
        notif = SMSNotificacao()
        notif.enviar(mensagem)
    elif tipo == "push":
        notif = PushNotificacao()
        notif.enviar(mensagem)
    elif tipo == "wpp":
        notif = WppNotificacao()
        notif.enviar(mensagem)
    else:
        print("Tipo não suportado")

# Uso
enviar_notificacao("email", "Bem-vindo!")
enviar_notificacao("sms", "Seu código é 1234")
enviar_notificacao("wpp","Oi, sumida!")
```

```python
# Classes de notificação
class Notificacao:
    def enviar(self, mensagem):
        pass

class EmailNotificacao(Notificacao):
    def enviar(self, mensagem):
        print(f"📧 Email enviado: {mensagem}")

class SMSNotificacao(Notificacao):
    def enviar(self, mensagem):
        print(f"📱 SMS enviado: {mensagem}")

class PushNotificacao(Notificacao):
    def enviar(self, mensagem):
        print(f"🔔 Push enviado: {mensagem}")

class WppNotificacao(Notificacao):
    def enviar(self, mensagem):
        print(f"🔔 Wpp enviado: {mensagem}")

# Factory
class NotificacaoFactory:
    @staticmethod
    def criar(tipo):
        tipos = {
            "email": EmailNotificacao,
            "sms": SMSNotificacao,
            "push": PushNotificacao,
            "wpp": WppNotificacao
        }
        
        classe = tipos.get(tipo.lower())
        if classe:
            return classe()
        else:
            raise ValueError(f"Tipo de notificação '{tipo}' não suportado")

# Uso
notificacao_email = NotificacaoFactory.criar("email")
#notificacao_email = EmailNotificacao()
notificacao_email.enviar("Bem-vindo ao sistema!")

notificacao_sms = NotificacaoFactory.criar("sms")
notificacao_sms.enviar("Seu código é 1234")

notific_wpp = NotificacaoFactory.criar("wpp")
notific_wpp.enviar("Oi, sumida!")
```
