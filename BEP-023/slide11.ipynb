```python
class Subject:
    def __init__(self):
        self._observers = []
    
    def adicionar(self, observer):
        self._observers.append(observer)
    
    def remover(self, observer):
        self._observers.remove(observer)
    
    def notificar(self, evento):
        for observer in self._observers:
            observer.atualizar(evento)

class Observer:
    def __init__(self, nome):
        self.nome = nome

    def atualizar(self, evento):
        print(f"Observer {self.nome} recebeu: {evento}")

# Uso
subject = Subject()
observer1 = Observer("observador 1")
observer2 = Observer("observador 2")

subject.adicionar(observer1)
subject.adicionar(observer2)
subject.notificar("Evento importante!")
```

```python
class EventManager:
    def __init__(self):
        self._listeners = {}
    
    def on(self, evento, callback):
        if evento not in self._listeners:
            self._listeners[evento] = []
        self._listeners[evento].append(callback)
    
    def emit(self, evento, dados=None):
        if evento in self._listeners:
            for callback in self._listeners[evento]:
                callback(dados)

# Uso
eventos = EventManager()

# Registrar listeners
eventos.on("usuario_login", lambda d: print(f"✅ Login: {d}"))
eventos.on("usuario_login", lambda d: print(f"📧 Email enviado para {d}"))
eventos.on("compra_realizada", lambda d: print(f"💰 Compra: R$ {d}"))

# Emitir eventos
eventos.emit("usuario_login", "joao@email.com")
eventos.emit("compra_realizada", 150.00)
```

```python
# 1. O APARELHO (Genérico)
class SistemaDeAlarme:
    def __init__(self):
        self.instrucoes = [] # Gaveta vazia
    
    # Aqui o cliente entrega o papel com a instrução
    def adicionar_ordem(self, funcao_instrucao):
        self.instrucoes.append(funcao_instrucao)
    
    # Quando o ladrão entra...
    def disparar(self):
        print("🚨 ALARME DISPARADO! Executando ordens...")
        # O aparelho pega cada papel na gaveta e executa
        for acao in self.instrucoes:
            acao() # Executa a função que foi guardada

# 2. AS INSTRUÇÕES (O que cada cliente quer)
def ligar_policia():
    print("📞 Alô, 190? Tem ladrão aqui!")

def soltar_cachorros():
    print("🐕 Quem soltou os cachorros?!")

# 3. O USO (Configurando o aparelho)
alarme = SistemaDeAlarme()

# Aqui está o pulo do gato:
# Não estamos chamando ligar_policia()...
# Estamos ENTREGANDO a funcionalidade para o alarme usar DEPOIS.
alarme.adicionar_ordem(ligar_policia)
alarme.adicionar_ordem(soltar_cachorros)

# ... tempo passa ... 3 da manhã ...
alarme.disparar()
```
