```python
# Componente base
class Texto:
    def __init__(self, conteudo):
        self._conteudo = conteudo
    
    def renderizar(self):
        return self._conteudo

# Decorator
class Formatador(Texto):
    def __init__(self, texto):
        self._texto = texto
    
    def renderizar(self):
        return self._texto.renderizar()

# Decorators concretos
class Negrito(Formatador):
    def renderizar(self):
        return f"<n>{self._texto.renderizar()}<n>"

class Italico(Formatador):
    def renderizar(self):
        return f"<i>{self._texto.renderizar()}<i>"

class Sublinhado(Formatador):
    def renderizar(self):
        return f"<s>{self._texto.renderizar()}<s>"

# Uso - combinando formatações
texto = Texto("Olá Mundo")

texto_negrito = Negrito(texto)
print(texto_negrito.renderizar())
# Olá Mundo

texto_completo = Sublinhado(Italico(Negrito(Texto("Python"))))
print(texto_completo.renderizar())
# Python
```

```python
# Componente base
class Stream:
    def ler(self):
        pass
    
    def escrever(self, dados):
        pass

# Componente concreto
class FileStream(Stream):
    def __init__(self, arquivo):
        self.arquivo = arquivo
    
    def ler(self):
        return f"Dados de {self.arquivo}"
    
    def escrever(self, dados):
        print(f"Escrevendo em {self.arquivo}: {dados}")

# Decorator
class StreamDecorator(Stream):
    def __init__(self, stream):
        self._stream = stream
    
    def ler(self):
        return self._stream.ler()
    
    def escrever(self, dados):
        self._stream.escrever(dados)

# Decorators concretos
class CompressaoStream(StreamDecorator):
    def escrever(self, dados):
        dados_comprimidos = f"[COMPRIMIDO] {dados}"
        self._stream.escrever(dados_comprimidos)
    
    def ler(self):
        dados = self._stream.ler()
        return dados.replace("[COMPRIMIDO] ", "")

class CriptografiaStream(StreamDecorator):
    def escrever(self, dados):
        dados_criptografados = f"[CRIPTOGRAFADO] {dados}"
        self._stream.escrever(dados_criptografados)
    
    def ler(self):
        dados = self._stream.ler()
        return dados.replace("[CRIPTOGRAFADO] ", "")

# Uso - combinando decorators
stream_base = FileStream("dados.txt")

# Apenas compressão
stream_comp = CompressaoStream(stream_base)
stream_comp.escrever("Dados importantes")

# Compressão + Criptografia
stream_seguro = CriptografiaStream(CompressaoStream(FileStream("seguro.txt")))
stream_seguro.escrever("Dados secretos")
```
