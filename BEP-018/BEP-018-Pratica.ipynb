```python
# Definição básica de uma classe
class Pessoa:
    pass

# Instanciação de um objeto
p1 = Pessoa()
print(type(p1))  # <class '__main__.Pessoa'>
print(p1)  # <__main__.Pessoa object at 0x...>

```

```python
# Adicionando o método construtor __init__
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

# Criando objetos (instâncias)
p1 = Pessoa("Ana", 25)
p2 = Pessoa("Bruno", 30)

print(f"{p1.nome} tem {p1.idade} anos")
print(f"{p2.nome} tem {p2.idade} anos")

```

```python
# Múltiplas instâncias da mesma classe
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = float(preco)

# Criando vários produtos
produto1 = Produto("Caderno", 20.00)
produto2 = Produto("Caneta", 2.50)
produto3 = Produto("Borracha", 1.00)

print(f"{produto1.nome}: R$ {produto1.preco:.2f}")
print(f"{produto2.nome}: R$ {produto2.preco:.2f}")
print(f"{produto3.nome}: R$ {produto3.preco:.2f}")

```

```python
# Classe completa: Estudante com múltiplos métodos
class Estudante:
    def __init__(self, nome, matricula, curso):
        self.nome = nome
        self.matricula = matricula
        self.curso = curso
        self.notas = []  # Lista vazia inicialmente
    
    def adicionar_nota(self, nota):
        self.notas.append(float(nota))
    
    def calcular_media(self):
        if len(self.notas) > 0:
            return sum(self.notas) / len(self.notas)
        return 0.0
    
    def situacao(self):
        media = self.calcular_media()
        if media >= 7.0:
            return "Aprovado"
        elif media >= 5.0:
            return "Recuperação"
        else:
            return "Reprovado"

est1 = Estudante("Ana", "2025A001", "Python")
est1.adicionar_nota(8.5)
est1.adicionar_nota(7.0)
est1.adicionar_nota(9.0)

print(f"{est1.nome}: Média {est1.calcular_media():.1f} - {est1.situacao()}")

```

```python
# Exercício 1: Classe Retangulo
class Retangulo:
    def __init__(self, largura, altura):
        self.largura = float(largura)
        self.altura = float(altura)
    
    def area(self):
        return self.largura * self.altura
    
    def perimetro(self):
        return 2 * (self.largura + self.altura)

r1 = Retangulo(5, 3)
r2 = Retangulo(4, 4)

print(f"Retângulo 1 - Área: {r1.area()}, Perímetro: {r1.perimetro()}")
print(f"Retângulo 2 - Área: {r2.area()}, Perímetro: {r2.perimetro()}")

```

```python
# Exercício 2: Classe Livro
class Livro:
    def __init__(self, titulo, autor, paginas):
        self.titulo = titulo
        self.autor = autor
        self.paginas = int(paginas)
    
    def descricao(self):
        return f"'{self.titulo}' por {self.autor} ({self.paginas} páginas)"

l1 = Livro("Python para Iniciantes", "João Silva", 200)
l2 = Livro("POO Avançada", "Maria Santos", 350)

print(l1.descricao())
print(l2.descricao())

```

```python
# Exercício 3: Classe Aluno
class Aluno:
    def __init__(self, nome, notas):
        self.nome = nome
        self.notas = [float(n) for n in notas]
    
    def media(self):
        return sum(self.notas) / len(self.notas)
    
    def situacao(self):
        media = self.media()
        return "Aprovado" if media >= 6 else "Reprovado"

a1 = Aluno("Ana", [7, 8, 6, 9])
a2 = Aluno("Bruno", [5, 4, 6, 5])

print(f"{a1.nome}: Média {a1.media():.1f} - {a1.situacao()}")
print(f"{a2.nome}: Média {a2.media():.1f} - {a2.situacao()}")

```

```python
# Exercício Avançado: Classe ContaBancaria
class ContaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.saldo = float(saldo_inicial)
    
    def depositar(self, valor):
        if valor > 0:
            self.saldo += valor
            return f"Depósito de R$ {valor:.2f} realizado."
        return "Valor inválido para depósito."
    
    def sacar(self, valor):
        if valor > 0 and valor <= self.saldo:
            self.saldo -= valor
            return f"Saque de R$ {valor:.2f} realizado."
        return "Saldo insuficiente ou valor inválido."
    
    def ver_saldo(self):
        return f"Saldo atual: R$ {self.saldo:.2f}"

conta = ContaBancaria("Ana")
print(conta.depositar(100))
print(conta.sacar(30))
print(conta.ver_saldo())

```

```python
# Exercício Avançado: Classe Produto
class Produto:
    def __init__(self, nome, preco, estoque):
        self.nome = nome
        self.preco = float(preco)
        self.estoque = int(estoque)
    
    def aplicar_desconto(self, percentual):
        fator = 1 - (percentual / 100)
        self.preco = round(self.preco * fator, 2)
        return f"Desconto aplicado. Novo preço: R$ {self.preco:.2f}"
    
    def vender(self, quantidade):
        if quantidade > 0 and quantidade <= self.estoque:
            self.estoque -= quantidade
            return f"Venda de {quantidade} unidade(s) realizada(s). Estoque: {self.estoque}"
        return "Estoque insuficiente ou quantidade inválida."
    
    def repor(self, quantidade):
        if quantidade > 0:
            self.estoque += quantidade
            return f"Reposição de {quantidade} unidade(s). Estoque: {self.estoque}"
        return "Quantidade inválida."

p = Produto("Notebook", 2500.00, 10)
print(p.aplicar_desconto(10))
print(p.vender(3))
print(p.repor(5))

```
